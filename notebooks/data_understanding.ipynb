Sales demand is highly skewed, with most product-store-day observations having low demand and a small number having very high demand. Therefore, the dataset contains substantial demand variability and outliers, which should be considered during feature engineering and model selection

```python
daily_sales = df2[sales_cols].sum().reset_index()

daily_sales.columns = ['d', 'sales']

daily_sales = daily_sales.merge(
    df1[['d', 'weekday']],
    on='d',
    how='left'
)
```

```python
weekday_sales = daily_sales.groupby('weekday')['sales'].mean()

plt.figure(figsize=(10, 5))
weekday_sales.plot(kind='bar')
plt.title('Average Sales by Weekday')
plt.xlabel('Weekday')
plt.ylabel('Average Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
df2['avg_sales'] = df2[sales_cols].mean(axis=1)

store_sales = df2.groupby('store_id')['avg_sales'].mean()

plt.figure(figsize=(8, 5))
store_sales.plot(kind='bar')
plt.title('Average Sales by Store')
plt.xlabel('Store')
plt.ylabel('Average Sales')
plt.xticks(rotation=45)
plt.show()
```

Final EDA Summary

The M5 dataset consists of three main datasets: calendar.csv, sales_train_validation.csv, and sell_prices.csv. The calendar dataset provides date, weekday, month, year, event, and SNAP information. The sales dataset contains historical daily unit sales for different products and stores. The sell prices dataset contains the selling price of products for different stores and Walmart weeks.

The target variable for the forecasting model will be daily product demand (sales). The original sales data is stored in a wide format using columns such as d_1, d_2, etc., so it will later be transformed into a long time-series format.

The main identifiers are id, item_id, dept_id, cat_id, store_id, state_id, d, and wm_yr_wk. Some identifiers such as product, department, category, store, and state can also be used as categorical ML features.

Potential ML features include calendar features such as weekday, month, year, events, and SNAP indicators; price-related features such as selling price and price changes; and time-series features such as lagged sales and rolling averages.

The main data-quality observations are the large number of missing values in event columns, which are expected because most days have no events; a highly right-skewed sales distribution; a large number of zero/low-sales observations; and the wide structure of the original sales dataset. These issues will be addressed during preprocessing and feature engineering.

```python
import pandas as pd

sales = pd.read_csv("../data/raw/sales_train_validation.csv")

id_cols = [
    "id",
    "item_id",
    "dept_id",
    "cat_id",
    "store_id",
    "state_id"
]

sales_long = sales.melt(
    id_vars=id_cols,
    var_name="d",
    value_name="sales"
)

print(sales_long.shape)
print(sales_long.head())
```

```python
calendar=pd.read_csv("../data/processed/calendar.csv")
calendar["date"]=pd.to_datetime(calendar["date"])
print(calendar.head())
print(calendar.columns)
```

```python
sales_long=sales_long.merge(
    calendar,
    on='d',
    how='left'
)
print(sales_long.shape)
print(sales_long.head())
```

```python
print(sales_long.shape)
print(sales_long[["d", "date", "sales"]].head())
```

```python
prices=pd.read_csv("../data/processed/sell_prices.csv")
print(prices.head())
print(prices.shape)
print(prices.columns)
```

```python
sales_long=sales_long.merge(
    prices,
    on=["store_id","item_id","wm_yr_wk"],
    how="left"
)
```

```python
print(sales_long.shape)
print(sales_long[[
    "item_id",
    "store_id",
    "d",
    "wm_yr_wk",
    "sales",
    "sell_price"
]].head(10))
```

```python
print(sales_long["sell_price"].isna().mean())
```

```python
missing_prices = sales_long["sell_price"].isna().sum()
total_rows = len(sales_long)

print("Missing prices:", missing_prices)
print("Total rows:", total_rows)
print("Missing percentage:", missing_prices / total_rows * 100)
```

```python
print(sales_long["sell_price"].describe())
```

```python
sales_long["year"] = sales_long["date"].dt.year
sales_long["month"] = sales_long["date"].dt.month
sales_long["week"] = sales_long["date"].dt.isocalendar().week
sales_long["day"] = sales_long["date"].dt.day
sales_long["dayofweek"] = sales_long["date"].dt.dayofweek
```

```python
print(sales_long[[
    "date",
    "year",
    "month",
    "week",
    "day",
    "dayofweek"
]].head())
```

```python
print("Duplicate rows:", sales_long.duplicated().sum())
```
